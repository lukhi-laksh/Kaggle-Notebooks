```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

## Import CSV And convert to DataFrame

```python
df = pd.read_csv('student_mental_health_hybrid_learning.csv')
```

## Check Null Value

```python
df.isnull().sum()
```

## fill missing PHQ-9 scores with median

```python
df['phq9_score'] = df['phq9_score'].fillna(df['phq9_score'].median())
```

## fill missing GAD-7 scores with median

```python
df['gad7_score'] = df['gad7_score'].fillna(df['gad7_score'].median())
```

## fill missing satisfaction values with median

```python
df['online_learning_satisfaction'] = df['online_learning_satisfaction'].fillna(
    df['online_learning_satisfaction'].median()
)
```

## fill missing CGPA values with median

```python
df['cgpa'] = df['cgpa'].fillna(df['cgpa'].median())
```

## check remaining null values

```python
print(df.isnull().sum())
```

# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='age', bins=10, kde=True)
plt.title(f'{plot_no}. Age Distribution of Students')
plt.xlabel('Age')
plt.ylabel('Number of Students')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='cgpa', bins=15, kde=True)
plt.title(f'{plot_no}. Distribution of Student CGPA')
plt.xlabel('CGPA')
plt.ylabel('Number of Students')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='daily_sleep_hours', bins=15, kde=True)
plt.axvline(df['daily_sleep_hours'].mean(), linestyle='--', label='Mean Sleep')
plt.title(f'{plot_no}. Distribution of Daily Sleep Hours')
plt.xlabel('Daily Sleep Hours')
plt.ylabel('Number of Students')
plt.legend()
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='daily_screen_time_hours', bins=15, kde=True)
plt.axvline(df['daily_screen_time_hours'].mean(), linestyle='--', label='Mean Screen Time')
plt.title(f'{plot_no}. Distribution of Daily Screen Time')
plt.xlabel('Daily Screen Time Hours')
plt.ylabel('Number of Students')
plt.legend()
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='phq9_score', bins=15, kde=True)
plt.title(f'{plot_no}. Distribution of PHQ-9 Depression Scores')
plt.xlabel('PHQ-9 Score')
plt.ylabel('Number of Students')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='gad7_score', bins=15, kde=True)
plt.title(f'{plot_no}. Distribution of GAD-7 Anxiety Scores')
plt.xlabel('GAD-7 Score')
plt.ylabel('Number of Students')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='perceived_stress_scale', bins=15, kde=True)
plt.title(f'{plot_no}. Distribution of Perceived Stress Levels')
plt.xlabel('Perceived Stress Scale')
plt.ylabel('Number of Students')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='year_of_study', y='cgpa')
plt.title(f'{plot_no}. CGPA Variation Across Years of Study')
plt.xlabel('Year of Study')
plt.ylabel('CGPA')
show_fig()
plot_no += 1
```

```python
major_order = df.groupby('major_category')['cgpa'].median().sort_values(ascending=False).index
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='major_category', y='cgpa', order=major_order)
plt.title(f'{plot_no}. Academic Performance Comparison Across Major Categories')
plt.xlabel('Major Category')
plt.ylabel('CGPA')
plt.xticks(rotation=30)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='daily_sleep_hours', y='cgpa', hue='perceived_stress_scale', size='weekly_study_hours', alpha=0.7)
plt.title(f'{plot_no}. Sleep Duration, Stress Level and Academic Performance')
plt.xlabel('Daily Sleep Hours')
plt.ylabel('CGPA')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.regplot(data=df, x='gad7_score', y='phq9_score', scatter_kws={'alpha':0.5}, line_kws={'linewidth':2})
plt.title(f'{plot_no}. Strong Association Between Anxiety and Depression Scores')
plt.xlabel('GAD-7 Anxiety Score')
plt.ylabel('PHQ-9 Depression Score')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='class_attendance_percent', y='cgpa', hue='year_of_study', palette='viridis', alpha=0.7)
plt.title(f'{plot_no}. Class Attendance and CGPA Across Years of Study')
plt.xlabel('Class Attendance Percentage')
plt.ylabel('CGPA')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='has_part_time_job', y='cgpa')
sns.stripplot(data=df, x='has_part_time_job', y='cgpa', alpha=0.3)
plt.title(f'{plot_no}. Academic Performance of Students With and Without Part-Time Jobs')
plt.xlabel('Has Part-Time Job')
plt.ylabel('CGPA')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='financial_stress', y='phq9_score')
plt.title(f'{plot_no}. Depression Scores Across Financial Stress Levels')
plt.xlabel('Financial Stress Level')
plt.ylabel('PHQ-9 Score')
show_fig()
plot_no += 1
```

```python
mental_health_avg = df.groupby('semester_type')[['phq9_score', 'gad7_score', 'perceived_stress_scale']].mean().reset_index()
mental_health_avg = mental_health_avg.melt(id_vars='semester_type', var_name='Mental Health Measure', value_name='Average Score')
fig = plt.figure(figsize=(10,6))
sns.barplot(data=mental_health_avg, x='semester_type', y='Average Score', hue='Mental Health Measure')
plt.title(f'{plot_no}. Average Mental Health Scores by Semester Type')
plt.xlabel('Semester Type')
plt.ylabel('Average Score')
plt.legend(title='Mental Health Measure')
show_fig()
plot_no += 1
```

```python
corr = df.select_dtypes(include='number').corr()
fig = plt.figure(figsize=(16,10))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title(f'{plot_no}. Correlation Heatmap of Academic, Lifestyle and Mental Health Factors')
show_fig()
plot_no += 1
```
