# Importing

## Import Library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

## Import CSV And convert to DataFrame

```python
df = pd.read_csv('messi_goals_assists_2008_2026.csv')
```

# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.lineplot(data=df, x='date', y='goals', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Goal-Scoring Trend Over Career')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='season', y='goals', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Total Goals Scored Across Seasons')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='venue', y='goals')
plt.title(f'{plot_no}. Goal Distribution by Home and Away Matches')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='competition', order=df['competition'].value_counts().index)
plt.title(f'{plot_no}. Match Goal Events by Competition')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='competition', y='goals', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Total Goals Contributed Across Competitions')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='career_goals', y='career_assists', hue='venue', size='goals', sizes=(40,200))
plt.title(f'{plot_no}. Career Goals and Assists Relationship')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='career_goals', y='career_goal_contributions', hue='competition')
plt.title(f'{plot_no}. Career Goals vs Total Goal Contributions')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='venue', y='goals', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Home vs Away Total Goal Production')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='venue', y='assists', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Home vs Away Assist Production')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='goals', discrete=True, kde=False)
plt.title(f'{plot_no}. Distribution of Goals Scored per Match')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='assists', discrete=True, kde=False)
plt.title(f'{plot_no}. Distribution of Assists per Match')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='stage', y='goals', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Goal Production Across Match Stages')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='club_or_country', y='goals', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Goals by Club or Country Representation')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='goals', y='assists', hue='venue', style='competition')
plt.title(f'{plot_no}. Match Goals vs Assists Relationship')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='opponent', y='goals', estimator='sum', errorbar=None, order=df.groupby('opponent')['goals'].sum().sort_values(ascending=False).head(10).index)
plt.title(f'{plot_no}. Top Opponents by Total Goals Scored')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='stadium', y='goals', estimator='sum', errorbar=None, order=df.groupby('stadium')['goals'].sum().sort_values(ascending=False).head(10).index)
plt.title(f'{plot_no}. Stadiums with the Highest Goal Production')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='country', y='goals', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Goal Production by Country')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.lineplot(data=df, x='date', y='career_goal_contributions', errorbar=None)
plt.title(f'{plot_no}. Career Goal Contributions Growth Over Time')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='career_assists', y='career_goal_contributions', hue='venue', size='goals', sizes=(40,200))
plt.title(f'{plot_no}. Career Assists vs Overall Goal Contributions')
show_fig()
plot_no += 1
```

```python
corr = df.select_dtypes(include='number').corr()

fig = plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title(f'{plot_no}. Correlation Heatmap of Career and Match Performance Metrics')
show_fig()
plot_no += 1
```
