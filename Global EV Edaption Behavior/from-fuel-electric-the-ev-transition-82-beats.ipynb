# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='age', hue='ev_adoption_likelihood', kde=True, multiple='stack')
plt.title(f'{plot_no}. Age Distribution Across EV Adoption Likelihood Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='ev_adoption_likelihood', y='annual_income', order=['Low', 'Medium', 'High'])
plt.title(f'{plot_no}. Annual Income Differences by EV Adoption Likelihood')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='city_type', hue='ev_adoption_likelihood', hue_order=['Low', 'Medium', 'High'])
plt.title(f'{plot_no}. EV Adoption Likelihood Across City Types')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.violinplot(data=df, x='ev_adoption_likelihood', y='daily_commute_km', order=['Low', 'Medium', 'High'], inner='quartile')
plt.title(f'{plot_no}. Daily Commute Patterns by EV Adoption Likelihood')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='ev_adoption_likelihood', y='weekly_travel_distance_km', order=['Low', 'Medium', 'High'])
plt.title(f'{plot_no}. Weekly Travel Distance and EV Adoption Interest')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='daily_commute_km', y='weekly_travel_distance_km', hue='ev_adoption_likelihood', size='annual_income', alpha=0.7)
plt.title(f'{plot_no}. Commute Distance, Weekly Travel and EV Adoption Relationship')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='current_vehicle_type', y='fuel_expense_per_month', estimator='mean', errorbar=None)
plt.title(f'{plot_no}. Average Monthly Fuel Expense by Current Vehicle Type')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='current_vehicle_type', y='vehicle_age_years')
plt.title(f'{plot_no}. Vehicle Age Distribution Across Vehicle Types')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.violinplot(data=df, x='ev_adoption_likelihood', y='nearest_charging_station_km', order=['Low', 'Medium', 'High'], inner='box')
plt.title(f'{plot_no}. Charging Station Proximity and EV Adoption Likelihood')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='home_charging_available', hue='ev_adoption_likelihood', hue_order=['Low', 'Medium', 'High'])
plt.xticks([0, 1], ['Not Available', 'Available'])
plt.title(f'{plot_no}. Impact of Home Charging Availability on EV Adoption')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='monthly_energy_consumption_kwh', y='monthly_charging_cost', hue='home_charging_available', alpha=0.7)
plt.title(f'{plot_no}. Energy Consumption Versus Monthly Charging Cost')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.regplot(data=df, x='electricity_cost_per_kwh', y='monthly_charging_cost', scatter_kws={'alpha': 0.5}, line_kws={'linewidth': 2})
plt.title(f'{plot_no}. Electricity Price Influence on Monthly Charging Cost')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='ev_adoption_likelihood', y='environmental_awareness_score', order=['Low', 'Medium', 'High'])
plt.title(f'{plot_no}. Environmental Awareness as a Driver of EV Adoption')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='ev_adoption_likelihood', y='technology_affinity_score', order=['Low', 'Medium', 'High'])
plt.title(f'{plot_no}. Technology Affinity Across EV Adoption Likelihood Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='ev_adoption_likelihood', y='government_incentive_awareness', order=['Low', 'Medium', 'High'])
plt.title(f'{plot_no}. Government Incentive Awareness and EV Adoption Interest')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='range_anxiety_score', y='battery_replacement_concern', hue='ev_adoption_likelihood', alpha=0.7)
plt.title(f'{plot_no}. Range Anxiety and Battery Concern Among Potential EV Buyers')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='previous_ev_experience', y='range_anxiety_score')
plt.xticks([0, 1], ['No Previous Experience', 'Previous Experience'])
plt.title(f'{plot_no}. Previous EV Experience and Reduction in Range Anxiety')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='previous_ev_experience', hue='ev_adoption_likelihood', hue_order=['Low', 'Medium', 'High'])
plt.xticks([0, 1], ['No Previous Experience', 'Previous Experience'])
plt.title(f'{plot_no}. Previous EV Experience and Future Adoption Likelihood')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
adoption_profile = df.groupby('ev_adoption_likelihood')[['environmental_awareness_score', 'technology_affinity_score', 'range_anxiety_score', 'battery_replacement_concern']].mean().reindex(['Low', 'Medium', 'High'])
sns.heatmap(adoption_profile, annot=True, fmt='.2f', cmap='YlGnBu')
plt.title(f'{plot_no}. Average Consumer Attitude Profile by EV Adoption Likelihood')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(20, 12))
corr = df.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title(f'{plot_no}. Correlation Heatmap of Numerical EV Adoption Factors')
show_fig()
plot_no += 1
```
