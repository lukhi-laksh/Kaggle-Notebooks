# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Price_USD', kde=True)
plt.title(f'{plot_no}. Distribution of Vehicle Prices and Market Concentration')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Sales_Volume', kde=True)
plt.title(f'{plot_no}. Distribution of Vehicle Sales Volume')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Sales_Classification', order=df['Sales_Classification'].value_counts().index)
plt.title(f'{plot_no}. Sales Classification Frequency and Category Imbalance')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.barplot(data=df, x='Model', y='Price_USD', estimator='mean', errorbar=None)
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Average Vehicle Price Comparison Across Models')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.boxplot(data=df, x='Model', y='Price_USD')
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Vehicle Price Variability and Outliers Across Models')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Year', y='Price_USD', hue='Sales_Classification')
plt.title(f'{plot_no}. Relationship Between Vehicle Year and Price by Sales Category')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Engine_Size_L', y='Price_USD', hue='Fuel_Type')
plt.title(f'{plot_no}. Engine Size and Vehicle Price Relationship by Fuel Type')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.barplot(data=df, x='Fuel_Type', y='Price_USD', estimator='mean', errorbar=None)
plt.title(f'{plot_no}. Average Vehicle Price Comparison Across Fuel Types')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.barplot(data=df, x='Transmission', y='Price_USD', estimator='mean', errorbar=None)
plt.title(f'{plot_no}. Average Vehicle Price by Transmission Type')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.barplot(data=df, x='Region', y='Sales_Volume', estimator='mean', errorbar=None)
plt.xticks(rotation=30)
plt.title(f'{plot_no}. Regional Comparison of Average Vehicle Sales Volume')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.boxplot(data=df, x='Region', y='Price_USD')
plt.xticks(rotation=30)
plt.title(f'{plot_no}. Regional Differences in Vehicle Pricing and Price Dispersion')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.countplot(data=df, x='Fuel_Type', hue='Sales_Classification')
plt.title(f'{plot_no}. Sales Classification Breakdown Across Fuel Types')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.countplot(data=df, x='Transmission', hue='Sales_Classification')
plt.title(f'{plot_no}. Sales Classification Patterns by Transmission Type')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.barplot(data=df, x='Model', y='Sales_Volume', estimator='mean', errorbar=None)
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Average Sales Volume Comparison Across Vehicle Models')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Price_USD', y='Sales_Volume', hue='Sales_Classification', alpha=0.7)
plt.title(f'{plot_no}. Vehicle Price Versus Sales Volume by Sales Classification')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.lineplot(data=df, x='Year', y='Price_USD', estimator='mean', errorbar=None, marker='o')
plt.title(f'{plot_no}. Average Vehicle Price Trend Across Model Years')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.lineplot(data=df, x='Year', y='Sales_Volume', estimator='mean', errorbar=None, marker='o')
plt.title(f'{plot_no}. Average Vehicle Sales Volume Trend Across Model Years')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Engine_Size_L', y='Sales_Volume', hue='Sales_Classification', alpha=0.7)
plt.title(f'{plot_no}. Engine Capacity Versus Sales Volume Across Sales Categories')
show_fig()
plot_no += 1
```

```python
corr = df.select_dtypes(include='number').corr()

fig = plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title(f'{plot_no}. Correlation Heatmap of Vehicle Attributes, Pricing and Sales')
show_fig()
plot_no += 1
```
