# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Customer_Age', bins=15, kde=True)
plt.title(f'{plot_no}. Customer Age Distribution Across Travel Bookings')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Gender')
plt.title(f'{plot_no}. Booking Distribution by Customer Gender')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, y='Customer_Country', order=df['Customer_Country'].value_counts().index)
plt.title(f'{plot_no}. Booking Volume by Customer Country')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, y='Customer_City', order=df['Customer_City'].value_counts().head(10).index)
plt.title(f'{plot_no}. Top 10 Customer Cities by Booking Volume')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Total_Trip_Cost', bins=30, kde=True)
plt.title(f'{plot_no}. Distribution of Total Trip Cost')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Booking_Status', y='Total_Trip_Cost')
plt.title(f'{plot_no}. Trip Cost Variation Across Booking Statuses')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Transportation_Type', y='Total_Trip_Cost')
plt.title(f'{plot_no}. Trip Cost Comparison by Transportation Type')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Transportation_Type', y='Total_Trip_Cost', estimator='mean')
plt.title(f'{plot_no}. Average Trip Cost by Transportation Type')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Meal_Plan', y='Total_Trip_Cost', estimator='mean')
plt.title(f'{plot_no}. Average Trip Cost Across Meal Plans')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Payment_Method', y='Total_Trip_Cost', estimator='mean')
plt.title(f'{plot_no}. Average Trip Cost by Payment Method')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Customer_Age', y='Total_Trip_Cost', hue='Transportation_Type')
plt.title(f'{plot_no}. Relationship Between Customer Age and Trip Cost')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Discount_Amount', y='Total_Trip_Cost', hue='Booking_Status')
plt.title(f'{plot_no}. Discount Amount Versus Total Trip Cost by Booking Status')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Customer_Rating', bins=10, kde=True)
plt.title(f'{plot_no}. Distribution of Customer Ratings')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Booking_Status', y='Customer_Rating')
plt.title(f'{plot_no}. Customer Rating Variation Across Booking Statuses')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Booking_Status', y='Customer_Rating', estimator='mean')
plt.title(f'{plot_no}. Average Customer Rating by Booking Status')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Cancellation_Status')
plt.title(f'{plot_no}. Booking Cancellation Status Distribution')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, y='Cancellation_Reason', order=df['Cancellation_Reason'].value_counts().index)
plt.title(f'{plot_no}. Most Frequent Reasons for Booking Cancellation')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Cancellation_Status', y='Discount_Amount')
plt.title(f'{plot_no}. Discount Amount Variation by Cancellation Status')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Customer_Rating', y='Total_Trip_Cost', hue='Booking_Status')
plt.title(f'{plot_no}. Relationship Between Customer Rating and Total Trip Cost')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
corr = df.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title(f'{plot_no}. Correlation Heatmap of Numerical Travel Booking Variables')
show_fig()
plot_no += 1
```
