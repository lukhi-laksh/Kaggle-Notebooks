# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Booking Value', kde=True, bins=30)
plt.title(f'{plot_no}. Distribution of Ride Booking Values')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Ride Distance', kde=True, bins=30)
plt.title(f'{plot_no}. Distribution of Ride Distances')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Avg VTAT', kde=True, bins=30)
plt.title(f'{plot_no}. Distribution of Average Vehicle Turnaround Time')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Avg CTAT', kde=True, bins=30)
plt.title(f'{plot_no}. Distribution of Average Customer Turnaround Time')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Driver Ratings', kde=True, bins=20, color='steelblue')
plt.title(f'{plot_no}. Distribution of Driver Ratings')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Customer Rating', kde=True, bins=20, color='seagreen')
plt.title(f'{plot_no}. Distribution of Customer Ratings')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Ride Distance', y='Booking Value', alpha=0.5)
sns.regplot(data=df, x='Ride Distance', y='Booking Value', scatter=False, color='red')
plt.title(f'{plot_no}. Relationship Between Ride Distance and Booking Value')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Avg VTAT', y='Avg CTAT', hue='Vehicle Type', alpha=0.7)
plt.title(f'{plot_no}. Vehicle Type Comparison of VTAT and CTAT')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Vehicle Type', y='Booking Value')
plt.title(f'{plot_no}. Booking Value Distribution Across Vehicle Types')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Vehicle Type', y='Ride Distance')
plt.title(f'{plot_no}. Ride Distance Variation Across Vehicle Types')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Vehicle Type', y='Booking Value', estimator='mean', errorbar=None)
plt.title(f'{plot_no}. Average Booking Value by Vehicle Type')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Vehicle Type', order=df['Vehicle Type'].value_counts().index)
plt.title(f'{plot_no}. Ride Volume Comparison Across Vehicle Types')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Payment Method', y='Booking Value')
plt.xticks(rotation=30)
plt.title(f'{plot_no}. Booking Value Differences Across Payment Methods')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
payment_counts = df['Payment Method'].value_counts()
plt.pie(payment_counts, labels=payment_counts.index, autopct='%1.1f%%', startangle=90)
plt.title(f'{plot_no}. Share of Rides by Payment Method')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
top_pickups = df['Pickup Location'].value_counts().head(15)
sns.barplot(x=top_pickups.values, y=top_pickups.index)
plt.title(f'{plot_no}. Top 15 Most Frequent Pickup Locations')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
top_drops = df['Drop Location'].value_counts().head(15)
sns.barplot(x=top_drops.values, y=top_drops.index)
plt.title(f'{plot_no}. Top 15 Most Frequent Drop Locations')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Driver Ratings', y='Customer Rating', hue='Vehicle Type', alpha=0.6)
plt.title(f'{plot_no}. Relationship Between Driver and Customer Ratings')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Ride Values Imputed', y='Booking Value')
plt.title(f'{plot_no}. Booking Value Comparison by Ride Imputation Status')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Avg VTAT', y='Booking Value', hue='Vehicle Type', alpha=0.6)
plt.title(f'{plot_no}. Vehicle Turnaround Time Versus Booking Value')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,8))
corr = df.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5, center=0)
plt.title(f'{plot_no}. Correlation Heatmap of Numerical Ride Features')
show_fig()
plot_no += 1
```
