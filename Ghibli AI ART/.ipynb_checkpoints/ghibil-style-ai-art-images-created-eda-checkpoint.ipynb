# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='trend_search_volume', kde=True)
plt.title(f'{plot_no}. Distribution of Trend Search Volume and Search Intensity')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='hours_since_trend_started', kde=True)
plt.title(f'{plot_no}. Distribution of Trend Age and Trend Lifecycle Patterns')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='social_media_mentions_count', kde=True)
plt.title(f'{plot_no}. Distribution of Social Media Mentions and Engagement Intensity')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='images_created_per_hour', kde=True)
plt.title(f'{plot_no}. Distribution of Image Creation Rates Across Trends')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='trend_search_volume', y='social_media_mentions_count', alpha=0.6)
plt.title(f'{plot_no}. Relationship Between Search Volume and Social Media Mentions')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.regplot(data=df, x='hours_since_trend_started', y='trend_search_volume', scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title(f'{plot_no}. Trend Age Versus Search Volume: Detecting Search Decline')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='hours_since_trend_started', y='social_media_mentions_count', hue='platform', alpha=0.7)
plt.title(f'{plot_no}. Trend Age and Social Mentions Across Social Platforms')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='platform', y='trend_search_volume')
plt.title(f'{plot_no}. Search Volume Distribution and Outliers by Platform')
plt.xticks(rotation=30)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='platform', y='social_media_mentions_count')
plt.title(f'{plot_no}. Social Media Engagement Comparison Across Platforms')
plt.xticks(rotation=30)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='platform', y='trend_search_volume', estimator='mean', errorbar=None)
plt.title(f'{plot_no}. Average Trend Search Volume by Social Media Platform')
plt.xticks(rotation=30)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='platform', y='images_created_per_hour', estimator='mean', errorbar=None)
plt.title(f'{plot_no}. Average Image Creation Rate Across Social Media Platforms')
plt.xticks(rotation=30)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='art_style_variant', y='social_media_mentions_count', estimator='mean', errorbar=None)
plt.title(f'{plot_no}. Average Social Media Mentions by Art Style Variant')
plt.xticks(rotation=25)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='featured_by_influencer', y='trend_search_volume')
plt.title(f'{plot_no}. Influencer Feature Status Versus Trend Search Volume')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='featured_by_influencer', y='social_media_mentions_count', estimator='mean', errorbar=None)
plt.title(f'{plot_no}. Influence of Influencer Features on Average Social Mentions')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='social_media_mentions_count', y='images_created_per_hour', hue='featured_by_influencer', alpha=0.7)
plt.title(f'{plot_no}. Social Engagement Versus Image Production by Influencer Feature Status')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='trend_search_volume', y='images_created_per_hour', hue='art_style_variant', alpha=0.7)
plt.title(f'{plot_no}. Search Demand Versus Image Production Across Art Styles')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.violinplot(data=df, x='platform', y='hours_since_trend_started', inner='quartile')
plt.title(f'{plot_no}. Trend Lifecycle Age Distribution Across Platforms')
plt.xticks(rotation=30)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='hours_since_trend_started', y='images_created_per_hour', hue='platform', alpha=0.7)
plt.title(f'{plot_no}. Trend Age Versus Image Creation Rate Across Platforms')
show_fig()
plot_no += 1
```

```python
corr = df.select_dtypes(include='number').corr()
fig = plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title(f'{plot_no}. Correlation Heatmap of Numerical Trend and Content Metrics')
show_fig()
plot_no += 1
```
