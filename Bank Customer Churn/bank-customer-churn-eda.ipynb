# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='NumOfProducts')
plt.title(f'{plot_no}. Customer Distribution Across Number of Products')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='NumOfProducts', hue='Geography')
plt.title(f'{plot_no}. Product Ownership Distribution Across Geographies')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='NumOfProducts', hue='Gender')
plt.title(f'{plot_no}. Number of Products by Customer Gender')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='NumOfProducts', y='CreditScore')
plt.title(f'{plot_no}. Credit Score Variation Across Product Groups')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='NumOfProducts', y='Age')
plt.title(f'{plot_no}. Age Distribution Across Product Groups')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='NumOfProducts', y='Balance')
plt.title(f'{plot_no}. Account Balance Variation Across Product Groups')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='NumOfProducts', y='Tenure')
plt.title(f'{plot_no}. Customer Tenure Across Different Product Groups')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='NumOfProducts', y='Balance', errorbar=None)
plt.title(f'{plot_no}. Average Account Balance by Number of Products')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='NumOfProducts', y='CreditScore', errorbar=None)
plt.title(f'{plot_no}. Average Credit Score by Number of Products')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='NumOfProducts', y='Age', errorbar=None)
plt.title(f'{plot_no}. Average Customer Age by Number of Products')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Geography', y='NumOfProducts', errorbar=None)
plt.title(f'{plot_no}. Average Product Count Across Geographies')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Gender', y='NumOfProducts', errorbar=None)
plt.title(f'{plot_no}. Average Number of Products by Gender')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Age', y='NumOfProducts', hue='Geography', alpha=0.6)
plt.title(f'{plot_no}. Relationship Between Customer Age and Product Count')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='CreditScore', y='NumOfProducts', hue='Geography', alpha=0.6)
plt.title(f'{plot_no}. Credit Score Relationship with Product Count')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Balance', y='NumOfProducts', hue='Geography', alpha=0.6)
plt.title(f'{plot_no}. Account Balance Relationship with Product Count')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Balance', hue='NumOfProducts', kde=True, element='step')
plt.title(f'{plot_no}. Account Balance Distribution Across Product Groups')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='CreditScore', hue='NumOfProducts', kde=True, element='step')
plt.title(f'{plot_no}. Credit Score Distribution Across Product Groups')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.violinplot(data=df, x='NumOfProducts', y='Age')
plt.title(f'{plot_no}. Customer Age Density Across Product Groups')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.violinplot(data=df, x='NumOfProducts', y='Balance')
plt.title(f'{plot_no}. Balance Density Across Product Groups')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,9))
corr = df.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title(f'{plot_no}. Correlation Heatmap of Customer Attributes')
show_fig()
plot_no += 1
```
