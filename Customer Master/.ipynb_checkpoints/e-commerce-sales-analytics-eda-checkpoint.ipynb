# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['customer_age'], kde=True)
plt.title(f'{plot_no}. Customer Age Distribution and Customer Demographic Concentration')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(x=df['customer_age'])
plt.title(f'{plot_no}. Customer Age Spread and Identification of Age Outliers')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['customer_acquisition_cost'], kde=True)
plt.title(f'{plot_no}. Customer Acquisition Cost Distribution and Cost Concentration')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(x=df['customer_acquisition_cost'])
plt.title(f'{plot_no}. Customer Acquisition Cost Variability and Outlier Detection')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='customer_age', y='customer_acquisition_cost', hue='gender')
plt.title(f'{plot_no}. Relationship Between Customer Age and Acquisition Cost by Gender')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.regplot(data=df, x='customer_age', y='customer_acquisition_cost')
plt.title(f'{plot_no}. Age-Based Trend in Customer Acquisition Cost')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='customer_segment', y='customer_acquisition_cost')
plt.title(f'{plot_no}. Acquisition Cost Comparison Across Customer Segments')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='customer_segment')
plt.title(f'{plot_no}. Customer Distribution Across Segments')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='gender', y='customer_acquisition_cost')
plt.title(f'{plot_no}. Acquisition Cost Comparison Between Genders')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='gender', hue='customer_segment')
plt.title(f'{plot_no}. Customer Segment Composition Across Genders')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='region', y='customer_acquisition_cost')
plt.title(f'{plot_no}. Regional Differences in Customer Acquisition Cost')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='region', hue='customer_segment')
plt.title(f'{plot_no}. Customer Segment Distribution Across Regions')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='customer_country', y='customer_acquisition_cost')
plt.title(f'{plot_no}. Acquisition Cost Variation Across Customer Countries')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='customer_country', hue='gender')
plt.title(f'{plot_no}. Gender Composition Across Customer Countries')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='customer_segment', y='customer_age', estimator='mean')
plt.title(f'{plot_no}. Average Customer Age Across Customer Segments')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='region', y='customer_age', estimator='mean')
plt.title(f'{plot_no}. Average Customer Age Across Geographic Regions')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='customer_age', y='customer_acquisition_cost', hue='customer_segment', style='gender')
plt.title(f'{plot_no}. Customer Age, Segment, and Acquisition Cost Relationship')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.violinplot(data=df, x='customer_segment', y='customer_acquisition_cost', hue='gender')
plt.title(f'{plot_no}. Acquisition Cost Distribution by Segment and Gender')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='customer_acquisition_cost', hue='customer_segment', kde=True, element='step')
plt.title(f'{plot_no}. Acquisition Cost Distribution Across Customer Segments')
show_fig()
plot_no += 1
```

```python
corr = df.select_dtypes(include='number').corr()

fig = plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title(f'{plot_no}. Correlation Heatmap of Customer Demographic and Acquisition Variables')
show_fig()
plot_no += 1
```
