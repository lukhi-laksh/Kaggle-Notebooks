# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='selling_price', bins=40, kde=True)
plt.title(f'{plot_no}. Distribution of Used Car Selling Prices')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='km_driven', bins=40, kde=True)
plt.title(f'{plot_no}. Distribution of Total Kilometers Driven')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='year', bins=25, kde=True)
plt.title(f'{plot_no}. Distribution of Used Cars by Manufacturing Year')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='km_driven', y='selling_price', alpha=0.5)
plt.title(f'{plot_no}. Relationship Between Driving Distance and Selling Price')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.regplot(data=df, x='year', y='selling_price', scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
plt.title(f'{plot_no}. Impact of Manufacturing Year on Selling Price')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='transmission', y='selling_price')
plt.title(f'{plot_no}. Price Premium of Automatic Versus Manual Cars')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='fuel', order=df['fuel'].value_counts().index)
plt.title(f'{plot_no}. Market Composition of Used Cars by Fuel Type')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='transmission', order=df['transmission'].value_counts().index)
plt.title(f'{plot_no}. Distribution of Manual and Automatic Used Cars')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='seller_type', order=df['seller_type'].value_counts().index)
plt.title(f'{plot_no}. Distribution of Cars Across Seller Categories')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='fuel', y='selling_price', estimator='mean', errorbar=None)
plt.title(f'{plot_no}. Average Resale Price Across Different Fuel Types')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='transmission', y='selling_price', estimator='median', errorbar=None)
plt.title(f'{plot_no}. Median Selling Price by Transmission Type')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.lineplot(data=df, x='year', y='selling_price', errorbar=None)
plt.title(f'{plot_no}. Trend of Average Used Car Prices Across Manufacturing Years')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.lineplot(data=df, x='year', y='km_driven', errorbar=None)
plt.title(f'{plot_no}. Changes in Average Driving Distance Across Car Manufacturing Years')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.violinplot(data=df, x='fuel', y='selling_price', inner='quartile')
plt.title(f'{plot_no}. Selling Price Distribution and Variability by Fuel Type')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.scatterplot(data=df, x='year', y='selling_price', hue='transmission', alpha=0.6)
plt.title(f'{plot_no}. Manufacturing Year and Resale Price by Transmission Type')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,6))
sns.scatterplot(data=df, x='km_driven', y='selling_price', hue='fuel', alpha=0.6)
plt.title(f'{plot_no}. Relationship Between Mileage and Price Across Fuel Types')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,8))
corr = df.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title(f'{plot_no}. Correlation Heatmap of Numerical Used Car Features')
show_fig()
plot_no += 1
```
