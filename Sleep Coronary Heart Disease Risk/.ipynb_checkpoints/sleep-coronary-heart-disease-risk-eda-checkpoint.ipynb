# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='chd_event')
plt.title(f'{plot_no}. Distribution of Coronary Heart Disease Events')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['age_years'], bins=30, kde=True)
plt.title(f'{plot_no}. Age Distribution of Participants')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['bmi'], bins=30, kde=True)
plt.title(f'{plot_no}. Body Mass Index Distribution')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='chd_event', y='age_years')
plt.title(f'{plot_no}. Age Comparison Between CHD and Non-CHD Groups')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='chd_event', y='bmi')
plt.title(f'{plot_no}. BMI Variation Across CHD Outcome')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='chd_event', y='fasting_glucose_mmol_L')
plt.title(f'{plot_no}. Fasting Glucose Levels by CHD Outcome')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='chd_event', y='total_cholesterol_mmol_L')
plt.title(f'{plot_no}. Cholesterol Levels Across CHD Status')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='chd_event', y='triglyceride_mg_dL')
plt.title(f'{plot_no}. Triglyceride Distribution by CHD Event')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='age_years', y='sbp_mmHg', hue='chd_event')
plt.title(f'{plot_no}. Age vs Systolic Blood Pressure by CHD Outcome')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='bmi', y='fasting_glucose_mmol_L', hue='chd_event')
plt.title(f'{plot_no}. BMI and Fasting Glucose Relationship by CHD Status')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='total_cholesterol_mmol_L', y='triglyceride_mg_dL', hue='chd_event')
plt.title(f'{plot_no}. Cholesterol and Triglyceride Relationship by CHD Outcome')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='sex', hue='chd_event')
plt.title(f'{plot_no}. CHD Occurrence Across Gender')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='family_history_cvd', hue='chd_event')
plt.title(f'{plot_no}. Family History of CVD and CHD Incidence')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='cigarette_smoking', hue='chd_event')
plt.xticks(rotation=20)
plt.title(f'{plot_no}. Smoking Habit and CHD Outcome')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='alcohol_drinking', hue='chd_event')
plt.xticks(rotation=20)
plt.title(f'{plot_no}. Alcohol Consumption Pattern by CHD Outcome')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='exercise_leisure', hue='chd_event')
plt.xticks(rotation=20)
plt.title(f'{plot_no}. Leisure Exercise Intensity and CHD Risk')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='sleep_quality', hue='chd_event')
plt.xticks(rotation=20)
plt.title(f'{plot_no}. Sleep Quality Distribution Across CHD Status')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='follow_up_years', y='age_years', hue='chd_event')
plt.title(f'{plot_no}. Follow-up Duration vs Age by CHD Outcome')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.violinplot(data=df, x='chd_event', y='sbp_mmHg')
plt.title(f'{plot_no}. Systolic Blood Pressure Distribution by CHD Status')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
corr = df.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title(f'{plot_no}. Correlation Heatmap of Numerical Health Indicators')
show_fig()
plot_no += 1
```
