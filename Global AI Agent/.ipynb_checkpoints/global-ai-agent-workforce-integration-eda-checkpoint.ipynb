# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Productivity_Gain_Percent', kde=True)
plt.title(f'{plot_no}. Distribution of Productivity Gains from AI Adoption')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Autonomous_Agents_Deployed', y='Productivity_Gain_Percent', hue='Company_Size', size='Human_Roles_Replaced')
plt.title(f'{plot_no}. Relationship Between Autonomous Agents and Productivity Gains')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Company_Size', y='Productivity_Gain_Percent')
plt.title(f'{plot_no}. Productivity Gain Across Company Sizes')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Industry', y='Productivity_Gain_Percent', estimator='mean')
plt.title(f'{plot_no}. Average Productivity Gain by Industry')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Industry', y='Avg_Agent_Cost_Per_Month_USD')
plt.xticks(rotation=30)
plt.title(f'{plot_no}. Monthly AI Agent Cost Variation Across Industries')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Avg_Agent_Cost_Per_Month_USD', y='Productivity_Gain_Percent', hue='Industry')
plt.title(f'{plot_no}. AI Agent Cost Versus Productivity Gain')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Human_Roles_Replaced', y='Human_Roles_Augmented', hue='Company_Size', size='Productivity_Gain_Percent')
plt.title(f'{plot_no}. Human Roles Replaced Versus Augmented by AI')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Primary_AI_Agent_Role', y='Productivity_Gain_Percent', estimator='mean')
plt.xticks(rotation=35)
plt.title(f'{plot_no}. Productivity Gain by Primary AI Agent Role')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Primary_AI_Agent_Role', y='Months_To_Positive_ROI')
plt.xticks(rotation=35)
plt.title(f'{plot_no}. Time to Positive ROI Across AI Agent Roles')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Months_To_Positive_ROI', y='Productivity_Gain_Percent', hue='Has_Strict_AI_Governance')
plt.title(f'{plot_no}. Positive ROI Timeline Versus Productivity Gain')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Has_Strict_AI_Governance', y='Cybersecurity_Incidents_YTD')
plt.title(f'{plot_no}. Cybersecurity Incidents Under Different AI Governance Policies')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Has_Strict_AI_Governance', y='Employee_Sentiment_Score_1_to_10', estimator='mean')
plt.title(f'{plot_no}. Employee Sentiment by AI Governance Approach')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Productivity_Gain_Percent', y='Employee_Sentiment_Score_1_to_10', hue='Has_Strict_AI_Governance', size='Human_Roles_Augmented')
plt.title(f'{plot_no}. Productivity Gains Versus Employee Sentiment')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Human_Roles_Replaced', y='Employee_Sentiment_Score_1_to_10', hue='Company_Size')
plt.title(f'{plot_no}. Human Role Replacement Versus Employee Sentiment')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Cybersecurity_Incidents_YTD', hue='Has_Strict_AI_Governance', kde=True, element='step')
plt.title(f'{plot_no}. Distribution of Cybersecurity Incidents by AI Governance')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Autonomous_Agents_Deployed', y='Cybersecurity_Incidents_YTD', hue='Has_Strict_AI_Governance', size='Company_Size')
plt.title(f'{plot_no}. Autonomous Agent Deployment Versus Cybersecurity Incidents')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Industry', y='Months_To_Positive_ROI', estimator='mean')
plt.xticks(rotation=30)
plt.title(f'{plot_no}. Average Time to Positive ROI Across Industries')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Autonomous_Agents_Deployed', y='Avg_Agent_Cost_Per_Month_USD', hue='Company_Size', size='Productivity_Gain_Percent')
plt.title(f'{plot_no}. AI Agent Deployment Scale Versus Monthly Operating Cost')
show_fig()
plot_no += 1
```

```python
corr = df.select_dtypes(include='number').corr()

fig = plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title(f'{plot_no}. Correlation Heatmap of AI Adoption and Business Performance Metrics')
show_fig()
plot_no += 1
```
