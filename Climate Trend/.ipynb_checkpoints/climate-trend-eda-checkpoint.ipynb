# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['avg_temperature_c'], bins=30, kde=True)
plt.title(f'{plot_no}. Distribution of Average Temperature')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['precipitation_mm'], bins=30, kde=True)
plt.title(f'{plot_no}. Distribution of Daily Precipitation')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['co2_ppm'], bins=30, kde=True)
plt.title(f'{plot_no}. Distribution of Atmospheric CO₂ Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['humidity_percent'], bins=30, kde=True)
plt.title(f'{plot_no}. Distribution of Humidity Percentage')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(x='region', y='avg_temperature_c', data=df)
plt.title(f'{plot_no}. Regional Comparison of Average Temperature')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(x='region', y='precipitation_mm', data=df)
plt.title(f'{plot_no}. Regional Variation in Daily Precipitation')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(x='region', y='humidity_percent', data=df)
plt.title(f'{plot_no}. Regional Distribution of Humidity Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(x='region', y='co2_ppm', data=df)
plt.title(f'{plot_no}. Regional Comparison of Atmospheric CO₂ Concentration')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='co2_ppm', y='avg_temperature_c', hue='region')
plt.title(f'{plot_no}. CO₂ Concentration vs Average Temperature by Region')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='humidity_percent', y='avg_temperature_c', hue='region')
plt.title(f'{plot_no}. Humidity Influence on Average Temperature')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='precipitation_mm', y='humidity_percent', hue='region')
plt.title(f'{plot_no}. Relationship Between Precipitation and Humidity')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='temperature_anomaly_c', y='avg_temperature_c', hue='region')
plt.title(f'{plot_no}. Temperature Anomaly Compared with Average Temperature')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='co2_ppm', y='temperature_anomaly_c', hue='region')
plt.title(f'{plot_no}. CO₂ Levels and Temperature Anomalies')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.violinplot(x='region', y='temperature_anomaly_c', data=df)
plt.title(f'{plot_no}. Distribution of Temperature Anomalies Across Regions')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.violinplot(x='region', y='co2_ppm', data=df)
plt.title(f'{plot_no}. Atmospheric CO₂ Distribution by Region')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(x='region', data=df)
plt.title(f'{plot_no}. Number of Records Available by Region')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.regplot(data=df, x='co2_ppm', y='avg_temperature_c', scatter_kws={'alpha':0.5})
plt.title(f'{plot_no}. Regression Trend Between CO₂ Levels and Temperature')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.regplot(data=df, x='precipitation_mm', y='humidity_percent', scatter_kws={'alpha':0.5})
plt.title(f'{plot_no}. Regression Analysis of Precipitation and Humidity')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.pairplot(df[['avg_temperature_c','precipitation_mm','co2_ppm','temperature_anomaly_c','humidity_percent']])
plt.suptitle(f'{plot_no}. Pairwise Relationships Among Climate Variables', y=1.02)
plt.show()
plot_no += 1
```

```python
corr = df[['avg_temperature_c','precipitation_mm','co2_ppm','temperature_anomaly_c','humidity_percent']].corr()

fig = plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title(f'{plot_no}. Correlation Heatmap of Climate Variables')
show_fig()
plot_no += 1
```
