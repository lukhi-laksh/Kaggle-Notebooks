# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['Mental_Health_Score'], kde=True)
plt.title(f'{plot_no}. Distribution of Mental Health Scores Across Students')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['Physical_Health_Score'], kde=True)
plt.title(f'{plot_no}. Distribution of Physical Health Scores Across Students')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Daily_Social_Media_Hours', y='Mental_Health_Score', hue='Gender')
plt.title(f'{plot_no}. Impact of Daily Social Media Use on Mental Health')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Daily_AI_Tool_Usage_Hours', y='Mental_Health_Score', hue='Education_Level')
plt.title(f'{plot_no}. Relationship Between AI Tool Usage and Mental Health')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Sleep_Hours', y='Mental_Health_Score', hue='Gender')
plt.title(f'{plot_no}. Relationship Between Sleep Duration and Mental Health')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Physical_Activity_Hours', y='Mental_Health_Score', hue='Gender')
plt.title(f'{plot_no}. Association Between Physical Activity and Mental Health')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Sleep_Hours', y='Physical_Health_Score', hue='Education_Level')
plt.title(f'{plot_no}. Relationship Between Sleep Duration and Physical Health')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Physical_Activity_Hours', y='Physical_Health_Score', hue='Gender')
plt.title(f'{plot_no}. Association Between Physical Activity and Physical Health')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Gender', y='Mental_Health_Score')
plt.title(f'{plot_no}. Mental Health Score Comparison Across Genders')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Education_Level', y='Mental_Health_Score')
plt.title(f'{plot_no}. Mental Health Differences Across Education Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Gender', y='Physical_Health_Score')
plt.title(f'{plot_no}. Physical Health Score Comparison Across Genders')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Education_Level', y='Physical_Health_Score')
plt.title(f'{plot_no}. Physical Health Differences Across Education Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Education_Level', y='Daily_Social_Media_Hours', estimator='mean')
plt.title(f'{plot_no}. Average Social Media Usage by Education Level')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Education_Level', y='Daily_AI_Tool_Usage_Hours', estimator='mean')
plt.title(f'{plot_no}. Average AI Tool Usage by Education Level')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Gender', y='Sleep_Hours', estimator='mean')
plt.title(f'{plot_no}. Average Sleep Duration Across Genders')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Gender', y='Physical_Activity_Hours', estimator='mean')
plt.title(f'{plot_no}. Average Physical Activity Across Genders')
show_fig()
plot_no += 1
```

```python
corr = df.select_dtypes(include='number').corr()
fig = plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title(f'{plot_no}. Correlation Heatmap of Student Lifestyle and Health Indicators')
show_fig()
plot_no += 1
```
