# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Funding_Stage', order=df['Funding_Stage'].value_counts().index)
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Distribution of Companies Across Funding Stages')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, y='Domain', order=df['Domain'].value_counts().index)
plt.title(f'{plot_no}. Startup Distribution by Business Domain')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['Total_Funding_USD_Millions'], bins=30, kde=True)
plt.title(f'{plot_no}. Distribution of Total Funding Raised')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['Valuation_USD_Millions'], bins=30, kde=True)
plt.title(f'{plot_no}. Distribution of Company Valuations')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Total_Funding_USD_Millions', y='Valuation_USD_Millions', hue='Funding_Stage')
plt.title(f'{plot_no}. Funding vs Valuation by Funding Stage')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Investor_Tier', y='Valuation_USD_Millions', estimator='mean')
plt.xticks(rotation=20)
plt.title(f'{plot_no}. Average Valuation by Investor Tier')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Monthly_Burn_Rate_Millions', y='Runway_Months_2024', hue='Funding_Stage')
plt.title(f'{plot_no}. Burn Rate vs Financial Runway')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Peak_Headcount_2023', y='Layoffs_2024_2025', hue='Acquisition_Status')
plt.title(f'{plot_no}. Layoffs Relative to Peak Workforce')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Peak_Headcount_2023', y='Current_Headcount_2026', hue='Funding_Stage')
plt.title(f'{plot_no}. Workforce Retention After Peak Hiring')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Acquisition_Status')
plt.title(f'{plot_no}. Distribution of Company Acquisition Status')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Funding_Stage', y='Runway_Months_2024', estimator='mean')
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Average Financial Runway by Funding Stage')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.violinplot(data=df, x='Funding_Stage', y='Monthly_Burn_Rate_Millions')
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Burn Rate Distribution Across Funding Stages')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Domain', y='Revenue_ARR_Millions', estimator='mean')
plt.xticks(rotation=90)
plt.title(f'{plot_no}. Average Revenue Across Startup Domains')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Founding_Year', y='Valuation_USD_Millions', hue='Funding_Stage')
plt.title(f'{plot_no}. Company Valuation by Founding Year')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
corr = df.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title(f'{plot_no}. Correlation Heatmap of Numerical Features')
show_fig()
plot_no += 1
```
