# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Primary Genre', order=df['Primary Genre'].value_counts().index)
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Distribution of Artists Across Primary Genres')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Country of Origin', order=df['Country of Origin'].value_counts().head(15).index)
plt.xticks(rotation=60)
plt.title(f'{plot_no}. Top Countries by Number of Artists')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['Total Streams (in millions)'], bins=30, kde=True)
plt.title(f'{plot_no}. Distribution of Total Streams')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Lead Streams (in millions)', y='Total Streams (in millions)', hue='Primary Genre')
plt.title(f'{plot_no}. Lead Streams vs Total Streams by Genre')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Debut Year', y='Total Streams (in millions)', hue='Sex')
plt.title(f'{plot_no}. Artist Debut Year vs Total Streams')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.regplot(data=df, x='Solo Streams (in millions)', y='Total Streams (in millions)', scatter_kws={'alpha':0.6})
plt.title(f'{plot_no}. Relationship Between Solo Streams and Total Streams')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.regplot(data=df, x='Collaborative Streams (in millions)', y='Total Streams (in millions)', scatter_kws={'alpha':0.6})
plt.title(f'{plot_no}. Relationship Between Collaborative Streams and Total Streams')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['% of Solo Streams'], bins=30, kde=True)
plt.title(f'{plot_no}. Distribution of Solo Stream Percentage')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['% of Collaborative Streams'], bins=30, kde=True)
plt.title(f'{plot_no}. Distribution of Collaborative Stream Percentage')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
top_genres = df.groupby('Primary Genre')['Total Streams (in millions)'].mean().sort_values(ascending=False)
sns.barplot(x=top_genres.index, y=top_genres.values)
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Average Total Streams by Genre')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
top_languages = df.groupby('Primary Language')['Total Streams (in millions)'].mean().sort_values(ascending=False)
sns.barplot(x=top_languages.index, y=top_languages.values)
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Average Total Streams by Primary Language')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Primary Genre', y='% of Solo Streams')
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Solo Stream Percentage Across Genres')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Primary Genre', y='% of Collaborative Streams')
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Collaborative Stream Percentage Across Genres')
show_fig()
plot_no += 1
```

```python
corr = df.select_dtypes(include='number').corr()

fig = plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title(f'{plot_no}. Correlation Heatmap of Numerical Features')
show_fig()
plot_no += 1
```
