# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='video_category', order=df['video_category'].value_counts().index)
plt.title(f'{plot_no}. Distribution of Videos Across Content Categories')
plt.xlabel('Video Category')
plt.ylabel('Number of Videos')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='engagement_level', order=df['engagement_level'].value_counts().index)
plt.title(f'{plot_no}. Overall Distribution of Viewer Engagement Levels')
plt.xlabel('Engagement Level')
plt.ylabel('Number of Viewers')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='channel_country', order=df['channel_country'].value_counts().index)
plt.title(f'{plot_no}. Countries Contributing the Highest Number of Videos')
plt.xlabel('Channel Country')
plt.ylabel('Number of Videos')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='content_language', order=df['content_language'].value_counts().index)
plt.title(f'{plot_no}. Most Frequently Used Content Languages')
plt.xlabel('Content Language')
plt.ylabel('Number of Videos')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='upload_day', order=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])
plt.title(f'{plot_no}. Weekly Video Upload Pattern by Day')
plt.xlabel('Upload Day')
plt.ylabel('Number of Videos')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='video_length_bucket', order=['<5 min', '5-10 min', '10-20 min', '20-40 min', '40+ min'])
plt.title(f'{plot_no}. Distribution of Videos by Length Category')
plt.xlabel('Video Length')
plt.ylabel('Number of Videos')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='traffic_source', order=df['traffic_source'].value_counts().index)
plt.title(f'{plot_no}. Traffic Sources Driving the Most Video Views')
plt.xlabel('Traffic Source')
plt.ylabel('Number of Viewers')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='device_type', order=df['device_type'].value_counts().index)
plt.title(f'{plot_no}. Viewer Distribution Across Device Types')
plt.xlabel('Device Type')
plt.ylabel('Number of Viewers')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='viewer_age_group', order=['13-17', '18-24', '25-34', '35-44', '45-54', '55+'])
plt.title(f'{plot_no}. Audience Distribution Across Viewer Age Groups')
plt.xlabel('Viewer Age Group')
plt.ylabel('Number of Viewers')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='subscriber_status', order=df['subscriber_status'].value_counts().index)
plt.title(f'{plot_no}. Comparison of Subscribed and Non-Subscribed Viewers')
plt.xlabel('Subscriber Status')
plt.ylabel('Number of Viewers')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='video_category', hue='engagement_level', order=df['video_category'].value_counts().index, hue_order=['Low', 'Medium', 'High', 'Very High'])
plt.title(f'{plot_no}. Engagement Level Differences Across Video Categories')
plt.xlabel('Video Category')
plt.ylabel('Number of Viewers')
plt.xticks(rotation=45)
plt.legend(title='Engagement Level')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='traffic_source', hue='engagement_level', order=df['traffic_source'].value_counts().index, hue_order=['Low', 'Medium', 'High', 'Very High'])
plt.title(f'{plot_no}. Viewer Engagement Generated by Each Traffic Source')
plt.xlabel('Traffic Source')
plt.ylabel('Number of Viewers')
plt.xticks(rotation=45)
plt.legend(title='Engagement Level')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='device_type', hue='engagement_level', order=df['device_type'].value_counts().index, hue_order=['Low', 'Medium', 'High', 'Very High'])
plt.title(f'{plot_no}. Engagement Performance Across Viewer Devices')
plt.xlabel('Device Type')
plt.ylabel('Number of Viewers')
plt.xticks(rotation=45)
plt.legend(title='Engagement Level')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='subscriber_status', hue='engagement_level', order=df['subscriber_status'].value_counts().index, hue_order=['Low', 'Medium', 'High', 'Very High'])
plt.title(f'{plot_no}. Impact of Subscriber Status on Viewer Engagement')
plt.xlabel('Subscriber Status')
plt.ylabel('Number of Viewers')
plt.legend(title='Engagement Level')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='video_length_bucket', hue='engagement_level', order=['<5 min', '5-10 min', '10-20 min', '20-40 min', '40+ min'], hue_order=['Low', 'Medium', 'High', 'Very High'])
plt.title(f'{plot_no}. Relationship Between Video Length and Engagement Level')
plt.xlabel('Video Length')
plt.ylabel('Number of Viewers')
plt.xticks(rotation=45)
plt.legend(title='Engagement Level')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='upload_day', hue='engagement_level', order=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'], hue_order=['Low', 'Medium', 'High', 'Very High'])
plt.title(f'{plot_no}. Viewer Engagement Patterns Across Upload Days')
plt.xlabel('Upload Day')
plt.ylabel('Number of Viewers')
plt.xticks(rotation=45)
plt.legend(title='Engagement Level')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
category_device = pd.crosstab(df['video_category'], df['device_type'], normalize='index') * 100
category_device.plot(kind='bar', stacked=True, ax=plt.gca())
plt.title(f'{plot_no}. Device Preference Within Each Video Category')
plt.xlabel('Video Category')
plt.ylabel('Percentage of Viewers')
plt.xticks(rotation=45)
plt.legend(title='Device Type', bbox_to_anchor=(1.02, 1), loc='upper left')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
corr = df.apply(lambda column: pd.factorize(column)[0]).corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title(f'{plot_no}. Correlation Heatmap of Encoded Video and Viewer Attributes')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
show_fig()
plot_no += 1
```
