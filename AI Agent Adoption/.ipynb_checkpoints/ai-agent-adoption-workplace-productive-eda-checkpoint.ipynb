# Importing

## Import Library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

## Import CSV And convert to DataFrame

```python
df = pd.read_csv('AI_Agent_Adoption_Workplace_Productivity_2024_2026.csv')
```

## Information of all Columns

```python
df.info()
```

## Check Null Value

```python
df.isnull().sum()
```

## Handle missing values in categorical columns

```python
cat_cols = df.select_dtypes(include='object').columns
df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])
```

## Handle missing values in numeric columns

```python
num_cols = df.select_dtypes(include='number').columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
```

## Check Dupicate Value

```python
df.duplicated().sum()
```

## Summary

```python
df.describe()
```

# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='AI_Agent_Usage', order=df['AI_Agent_Usage'].value_counts().index)
plt.title(f'{plot_no}. Distribution of AI Agent Usage Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='AI_Agent_Usage', y='Productivity_Change_Percent')
plt.title(f'{plot_no}. Productivity Change Across AI Agent Usage Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Hours_Saved_Per_Week', y='Productivity_Change_Percent', hue='AI_Agent_Usage')
plt.title(f'{plot_no}. Hours Saved vs Productivity Change')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.regplot(data=df, x='Hours_Saved_Per_Week', y='Productivity_Change_Percent', scatter_kws={'alpha':0.5})
plt.title(f'{plot_no}. Relationship Between Hours Saved and Productivity Change')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Tasks_Automated', y='Hours_Saved_Per_Week', hue='AI_Agent_Usage')
plt.title(f'{plot_no}. Automated Tasks vs Weekly Hours Saved')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='AI_Skill_Level', y='Productivity_Change_Percent', order=['Beginner','Intermediate','Advanced'])
plt.title(f'{plot_no}. Productivity Change Across AI Skill Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='AI_Skill_Level', y='Hours_Saved_Per_Week', order=['Beginner','Intermediate','Advanced'], estimator='mean')
plt.title(f'{plot_no}. Average Weekly Hours Saved by AI Skill Level')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Job_Risk_Level', y='Trust_in_AI')
plt.title(f'{plot_no}. Trust in AI Across Job Risk Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Trust_in_AI', y='Job_Satisfaction', hue='Human_AI_Collaboration')
plt.title(f'{plot_no}. Trust in AI vs Job Satisfaction by Collaboration Level')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Training_Hours', y='Productivity_Change_Percent', hue='AI_Skill_Level')
plt.title(f'{plot_no}. Training Hours vs Productivity Change')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Workload_Level', y='Error_Rate', order=['Low','Moderate','High'])
plt.title(f'{plot_no}. Error Rate Across Workload Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Error_Rate', y='Job_Satisfaction', hue='Workload_Level')
plt.title(f'{plot_no}. Error Rate vs Job Satisfaction by Workload')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Overall_AI_Impact', y='Productivity_Change_Percent')
plt.title(f'{plot_no}. Productivity Change Across Overall AI Impact')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Overall_AI_Impact', hue='AI_Agent_Usage')
plt.title(f'{plot_no}. AI Impact Distribution Across Usage Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Industry', y='Productivity_Change_Percent', estimator='mean')
plt.xticks(rotation=45, ha='right')
plt.title(f'{plot_no}. Average Productivity Change Across Industries')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Company_Size', y='Hours_Saved_Per_Week', estimator='mean')
plt.xticks(rotation=30, ha='right')
plt.title(f'{plot_no}. Weekly Hours Saved Across Company Sizes')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='AI_Agent_Type', y='Productivity_Change_Percent')
plt.xticks(rotation=45, ha='right')
plt.title(f'{plot_no}. Productivity Change by AI Agent Type')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Monthly_AI_Cost', y='Productivity_Change_Percent', hue='Overall_AI_Impact')
plt.title(f'{plot_no}. Monthly AI Cost vs Productivity Change')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,8))
corr = df.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title(f'{plot_no}. Correlation Heatmap of Numerical AI Workforce Metrics')
show_fig()
plot_no += 1
```
