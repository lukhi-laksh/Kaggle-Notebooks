# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['Total_Price'], kde=True)
plt.title(f'{plot_no}. Distribution of Order Total Price')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['Delivery_Duration_Minutes'], kde=True)
plt.title(f'{plot_no}. Distribution of Delivery Duration')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Delivery_Distance_km', y='Delivery_Duration_Minutes', hue='Traffic_Level')
plt.title(f'{plot_no}. Delivery Distance vs Duration Across Traffic Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Quantity', y='Total_Price', hue='Item_Name')
plt.title(f'{plot_no}. Quantity vs Total Price by Food Item')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Traffic_Level', y='Delivery_Duration_Minutes')
plt.title(f'{plot_no}. Delivery Duration Differences Across Traffic Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Driver_Availability', y='Delivery_Duration_Minutes')
plt.title(f'{plot_no}. Delivery Duration by Driver Availability')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Driver_Vehicle', y='Delivery_Duration_Minutes')
plt.title(f'{plot_no}. Delivery Duration by Driver Vehicle Type')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Payment_Method', y='Total_Price', estimator='mean')
plt.title(f'{plot_no}. Average Order Value by Payment Method')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Order_Status', y='Delivery_Duration_Minutes', estimator='mean')
plt.title(f'{plot_no}. Average Delivery Duration by Order Status')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Traffic_Level', y='Delivery_Distance_km', estimator='mean')
plt.title(f'{plot_no}. Average Delivery Distance Across Traffic Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='City', y='Total_Price', estimator='mean')
plt.title(f'{plot_no}. Average Order Value Across Cities')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='City', y='Delivery_Duration_Minutes', estimator='mean')
plt.title(f'{plot_no}. Average Delivery Duration Across Cities')
plt.xticks(rotation=45)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Traffic_Level', hue='Order_Status')
plt.title(f'{plot_no}. Order Status Distribution Across Traffic Conditions')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Driver_Vehicle', hue='Traffic_Level')
plt.title(f'{plot_no}. Driver Vehicle Usage Across Traffic Levels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Payment_Method', hue='Order_Status')
plt.title(f'{plot_no}. Order Status Patterns by Payment Method')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Total_Price', y='Delivery_Duration_Minutes', hue='Traffic_Level', size='Quantity')
plt.title(f'{plot_no}. Order Value vs Delivery Duration by Traffic and Quantity')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Delivery_Distance_km', y='Total_Price', hue='Payment_Method')
plt.title(f'{plot_no}. Delivery Distance vs Order Value by Payment Method')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.violinplot(data=df, x='Traffic_Level', y='Delivery_Duration_Minutes', inner='quartile')
plt.title(f'{plot_no}. Delivery Duration Distribution by Traffic Intensity')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
temp_df = df.copy()
temp_df['Order_Hour'] = pd.to_datetime(temp_df['Order_Time'], dayfirst=True).dt.hour
sns.lineplot(data=temp_df, x='Order_Hour', y='Delivery_Duration_Minutes', estimator='mean', errorbar=None, marker='o')
plt.title(f'{plot_no}. Average Delivery Duration by Order Hour')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(12,8))
corr = df.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title(f'{plot_no}. Correlation Heatmap of Numerical Order and Delivery Metrics')
show_fig()
plot_no += 1
```
