# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['Confirmed'], bins=30, kde=True)
plt.title(f'{plot_no}. Distribution of Confirmed COVID-19 Cases')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['Recovered'], bins=30, kde=True, color='green')
plt.title(f'{plot_no}. Distribution of Recovered Cases')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['Deaths'], bins=30, kde=True, color='red')
plt.title(f'{plot_no}. Distribution of COVID-19 Deaths')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Confirmed', y='Recovered', alpha=0.7)
plt.title(f'{plot_no}. Relationship Between Confirmed and Recovered Cases')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Confirmed', y='Deaths', alpha=0.7, color='red')
plt.title(f'{plot_no}. Impact of Confirmed Cases on Death Count')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Tests_Conducted', y='Confirmed', alpha=0.7)
plt.title(f'{plot_no}. Testing Volume vs Confirmed Cases')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Vaccination_Rate_%', y='Confirmed', alpha=0.7)
plt.title(f'{plot_no}. Vaccination Rate vs Confirmed Cases')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Vaccination_Rate_%', y='Deaths', alpha=0.7, color='purple')
plt.title(f'{plot_no}. Vaccination Rate vs Death Count')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Vaccination_Rate_%', y='Recovered', alpha=0.7, color='green')
plt.title(f'{plot_no}. Vaccination Rate vs Recovery Count')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Country', y='Confirmed')
plt.xticks(rotation=90)
plt.title(f'{plot_no}. Country-wise Distribution of Confirmed Cases')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Country', y='Deaths')
plt.xticks(rotation=90)
plt.title(f'{plot_no}. Country-wise Distribution of Deaths')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Country', y='Vaccination_Rate_%')
plt.xticks(rotation=90)
plt.title(f'{plot_no}. Country-wise Vaccination Rate Distribution')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
country_confirmed = df.groupby('Country')['Confirmed'].mean().sort_values(ascending=False)
sns.barplot(x=country_confirmed.index, y=country_confirmed.values)
plt.xticks(rotation=90)
plt.title(f'{plot_no}. Average Confirmed Cases by Country')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
country_deaths = df.groupby('Country')['Deaths'].mean().sort_values(ascending=False)
sns.barplot(x=country_deaths.index, y=country_deaths.values)
plt.xticks(rotation=90)
plt.title(f'{plot_no}. Average Death Count by Country')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
country_vacc = df.groupby('Country')['Vaccination_Rate_%'].mean().sort_values(ascending=False)
sns.barplot(x=country_vacc.index, y=country_vacc.values)
plt.xticks(rotation=90)
plt.title(f'{plot_no}. Average Vaccination Rate by Country')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
daily_confirmed = df.groupby('Date')['Confirmed'].sum()
plt.plot(daily_confirmed.index, daily_confirmed.values, marker='o')
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Daily Trend of Confirmed Cases')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
daily_deaths = df.groupby('Date')['Deaths'].sum()
plt.plot(daily_deaths.index, daily_deaths.values, marker='o', color='red')
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Daily Trend of COVID-19 Deaths')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
daily_vacc = df.groupby('Date')['Vaccination_Rate_%'].mean()
plt.plot(daily_vacc.index, daily_vacc.values, marker='o', color='green')
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Daily Average Vaccination Rate')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.pairplot(df[['Confirmed','Recovered','Deaths','Active','Tests_Conducted','Vaccination_Rate_%']])
plt.suptitle(f'{plot_no}. Pairwise Relationships Among Key COVID-19 Metrics', y=1.02)
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
corr = df.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title(f'{plot_no}. Correlation Heatmap of COVID-19 Indicators')
show_fig()
plot_no += 1
```
