# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Sales_Amount', kde=True, bins=30)
plt.title(f'{plot_no}. Distribution of Sales Amount with Density Pattern')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Sales_Amount', color='skyblue')
plt.title(f'{plot_no}. Sales Amount Distribution and Outlier Detection')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Region', y='Sales_Amount', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Total Sales Revenue Contribution Across Regions')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Sales_Rep', y='Sales_Amount', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Sales Representative Performance by Total Revenue')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Product_Category', y='Sales_Amount', estimator='mean', errorbar=None)
plt.title(f'{plot_no}. Average Sales Amount Across Product Categories')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Product_Category', y='Sales_Amount')
plt.title(f'{plot_no}. Sales Variability and Outliers Across Product Categories')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Quantity_Sold', y='Sales_Amount', hue='Product_Category', alpha=0.7)
plt.title(f'{plot_no}. Relationship Between Quantity Sold and Sales Revenue')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.regplot(data=df, x='Discount', y='Sales_Amount', scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title(f'{plot_no}. Impact of Discount Percentage on Sales Revenue')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Unit_Cost', y='Unit_Price', hue='Product_Category', alpha=0.7)
plt.title(f'{plot_no}. Unit Cost Versus Selling Price Across Product Categories')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Product_Category', y='Quantity_Sold', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Total Product Demand Across Different Categories')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Customer_Type', hue='Product_Category')
plt.title(f'{plot_no}. Customer Type Distribution Across Product Categories')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Customer_Type', y='Sales_Amount', estimator='mean', errorbar=None)
plt.title(f'{plot_no}. Average Spending Comparison Between New and Returning Customers')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Payment_Method', y='Sales_Amount', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Revenue Contribution by Payment Method')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Sales_Channel', y='Sales_Amount', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Revenue Comparison Between Online and Retail Sales Channels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Region', y='Quantity_Sold', hue='Sales_Channel', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Regional Product Demand Across Sales Channels')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Sales_Rep', y='Sales_Amount', hue='Region', estimator='sum', errorbar=None)
plt.title(f'{plot_no}. Sales Representative Revenue Performance by Region')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
df['Sale_Date'] = pd.to_datetime(df['Sale_Date'])
monthly_sales = df.set_index('Sale_Date').resample('ME')['Sales_Amount'].sum().reset_index()
sns.lineplot(data=monthly_sales, x='Sale_Date', y='Sales_Amount', marker='o')
plt.title(f'{plot_no}. Monthly Sales Revenue Trends Over Time')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Discount', y='Quantity_Sold', hue='Customer_Type', alpha=0.7)
plt.title(f'{plot_no}. Relationship Between Discounts and Customer Purchase Quantity')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Region_and_Sales_Rep', y='Sales_Amount', estimator='sum', errorbar=None)
plt.xticks(rotation=45, ha='right')
plt.title(f'{plot_no}. Combined Regional and Sales Representative Revenue Analysis')
show_fig()
plot_no += 1
```

```python
corr = df.select_dtypes(include='number').corr()

fig = plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title(f'{plot_no}. Correlation Heatmap of Numerical Sales Variables')
show_fig()
plot_no += 1
```
