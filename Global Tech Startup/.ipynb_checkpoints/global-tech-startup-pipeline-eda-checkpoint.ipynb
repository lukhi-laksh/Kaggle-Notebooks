# EDA

```python
def show_fig():
    plt.tight_layout()
    plt.show()

plot_no = 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Funding_Stage', order=df['Funding_Stage'].value_counts().index)
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Distribution of Startups Across Funding Stages')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Domain', order=df['Domain'].value_counts().index)
plt.xticks(rotation=90)
plt.title(f'{plot_no}. Startup Distribution by Business Domain')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['Valuation_USD_Millions'], bins=30, kde=True)
plt.title(f'{plot_no}. Distribution of Startup Valuations')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.histplot(df['Revenue_ARR_Millions'], bins=30, kde=True)
plt.title(f'{plot_no}. Annual Recurring Revenue Distribution')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Total_Funding_USD_Millions', y='Valuation_USD_Millions', hue='Funding_Stage')
plt.title(f'{plot_no}. Funding vs Valuation Across Funding Stages')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Revenue_ARR_Millions', y='Valuation_USD_Millions', hue='Funding_Stage')
plt.title(f'{plot_no}. Revenue Impact on Startup Valuation')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Monthly_Burn_Rate_Millions', y='Runway_Months_2024', hue='Funding_Stage')
plt.title(f'{plot_no}. Burn Rate vs Financial Runway')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
country_counts = df['Country'].value_counts()
plt.pie(country_counts.values, labels=country_counts.index, autopct='%1.1f%%', startangle=90)
plt.title(f'{plot_no}. Startup Distribution by Country')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df.groupby('Country', as_index=False)['Valuation_USD_Millions'].mean().sort_values('Valuation_USD_Millions', ascending=False),
            x='Country', y='Valuation_USD_Millions')
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Average Startup Valuation by Country')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.barplot(data=df.groupby('Domain', as_index=False)['Revenue_ARR_Millions'].mean().sort_values('Revenue_ARR_Millions', ascending=False),
            x='Domain', y='Revenue_ARR_Millions')
plt.xticks(rotation=90)
plt.title(f'{plot_no}. Average Revenue by Startup Domain')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Peak_Headcount_2023', y='Current_Headcount_2026', hue='Layoffs_2024_2025')
plt.title(f'{plot_no}. Workforce Retention After Layoffs')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Founding_Year', y='Valuation_USD_Millions', hue='Funding_Stage')
plt.title(f'{plot_no}. Startup Age vs Company Valuation')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.violinplot(data=df, x='Funding_Stage', y='Runway_Months_2024')
plt.xticks(rotation=45)
plt.title(f'{plot_no}. Runway Distribution Across Funding Stages')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Acquisition_Status', order=df['Acquisition_Status'].value_counts().index)
plt.xticks(rotation=20)
plt.title(f'{plot_no}. Acquisition Status of Startups')
show_fig()
plot_no += 1
```

```python
fig = plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Total_Funding_USD_Millions', y='Revenue_ARR_Millions', hue='Investor_Tier')
plt.title(f'{plot_no}. Funding Efficiency Based on Investor Tier')
show_fig()
plot_no += 1
```

```python
corr = df.select_dtypes(include='number').corr()

fig = plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title(f'{plot_no}. Correlation Heatmap of Numerical Startup Metrics')
show_fig()
plot_no += 1
```
